```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use ('dark_background')
```

```python
#import data
df = pd. read_csv(r"C:\Users\RADHIKA\Downloads\archive\zomato.csv")
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df = df.drop(['url','address','phone','menu_item','dish_liked','reviews_list'] , axis=1)
df.head()
```

```python
df.info()
```

```python
df.info()
```

```python
#removing duplicates
df.drop_duplicates(inplace = True )
df.shape
```

```python
df['rate'].unique()
```

```python
#removing "NEW",and "/5" from rate column
def handlerate (value):
    if (value == 'NEW' or value ==  '-'):
        return np.nan
    else:
        value = str(value).split('/')
        value = value [0]
        return float(value)
df['rate'] =df['rate'] .apply (handlerate)
df['rate'].head()
        
```

```python
#filling null values in rate column with mean
df['rate'].fillna(df['rate'].mean(),inplace = True)
```

```python
df['rate'].isnull().sum()
```

```python
df.info()
```

```python
#dropping null values
df.dropna(inplace = True)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
#renaming a column 
df.rename(columns= {'approx_cost(for two people)':'cost2plates'}, inplace = True)
df.head()
```

```python
df ['location'].unique()
```

```python
df['listed_in(city)'].unique()
```

```python
#drop listed city as both have same value and same work
df = df.drop(['listed_in(city)'], axis = 1
```

```python
df.head()
```

```python
df['cost2plates'].unique()
```

```python
#remove comma from cost2plates
def handlecomma(value):
    value = str(value)
    if ',' in value:
        value = value.replace(',','')
        return float (value)
    else:
        return float(value)
df['cost2plates'] = df['cost2plates'].apply(handlecomma)
df['cost2plates'].unique
        
```

```python
df.head()
```

```python
#cleaning rest type 
rest_type = df['rest_type'].value_counts(ascending = False)
rest_type
```

```python
rest_type_lessthan1000 = rest_type[rest_type<1000]
rest_type_lessthan1000
```

```python
# if data is less than 1000 in rest type , sum them all and count as 0
def handle_rest_type(value):
    if (value in rest_type_lessthan1000):
        return 'others'
    else:
        return value
df['rest_type'] = df['rest_type'].apply(handle_rest_type)
df['rest_type'].value_counts()
```

```python
location =df['location'].value_counts(ascending = False)
location_lessthan300 = location[location<300]
def handle_location(value):
    if(value  in location_lessthan300):
        return 'others'
    else:
        return value
df['location'] = df['location'].apply (handle_location)
df['location'].value_counts()
```

```python
#cleaning cuisines
cuisines = df['cuisines'].value_counts(ascending = False)
cuisines_lessthan100 = cuisines[cuisines<100]
def handle_cuisines(value):
    if(value in cuisines_lessthan100):
        return 'others'
    else:
        return value
df['cuisines'] = df['cuisines'].apply(handle_cuisines)
df['cuisines'].value_counts()
```
